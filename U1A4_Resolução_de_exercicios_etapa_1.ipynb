```python
from math import sqrt
a = 1
b = -5
c = 6

determinante = (b**2) - (4*a*c)

raiz = sqrt(determinante)
x1 =( (-b) + raiz)/(2*a)

x2 =( (-b) - raiz)/(2*a)

print("Determinante: ", determinante)
print(f"Valor de x1: {x1} e valor de x2: {x2}")
```
