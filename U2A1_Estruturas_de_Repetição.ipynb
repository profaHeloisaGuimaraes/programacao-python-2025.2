Escreva um programa que leia um número inteiro positivo N e exiba todos os números pares de 1 até N, além de mostrar a quantidade total de pares encontrados.

```python
#Resolução básica

num = int(input("Digite um valor inteiro: "))

contador = 0

for i in range (1, num+1, 2):
  print (i+1)
  contador +=1
print("Quantidade de números: ", contador)
```

```python
#Resolução indicada

num = int(input("Digite um valor inteiro: "))

contador = 0

for i in range (1, num+1):
  if i%2==0:
    print (i)
    contador +=1
print("Quantidade de números: ", contador)
```

Escreva um programa que leia dois números inteiros positivos, N e M, e calcule a soma de todos os múltiplos de M que sejam menores ou iguais a N.

```python
#Resolução usando FOR
num = int(input("Digite um valor limite: "))

multiplo = int(input("Digite o valor do multiplo: "))

soma = 0

for i in range(num+1):
  if (i % multiplo == 0):
    soma += i
print("Resultado da soma dos multiplos é ", soma)
```

```python
#Resolução usando WHILE

num = int(input("Digite um valor limite: "))

multiplo = int(input("Digite o valor do multiplo: "))

soma = 0

while num>0:
  if num % multiplo == 0:
    soma+=num
  num-=1

print("Resultado da soma dos multiplos é ", soma)
```
