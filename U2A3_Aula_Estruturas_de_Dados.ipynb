Crie uma lista de dicionários, onde cada dicionário representa um aluno com as chaves nome e nota. A lista deve ter três alunos. Em seguida, itere sobre a lista e imprima o nome de cada aluno que teve nota maior ou igual a 7.0.


```python
alunos = []

for cont in range (3):
  nome_aluno=input("Digite o nome: ")
  nota_aluno=int(input("Digite a nota: "))

  alunos.append({"Nome": nome_aluno, "Nota": nota_aluno})

print(alunos)

for item in alunos:
  if item["Nota"]>=7:
    print(item["Nome"])
```
